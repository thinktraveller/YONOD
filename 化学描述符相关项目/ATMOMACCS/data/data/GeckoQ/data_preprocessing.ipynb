```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('Dataframe.csv')
df.head(3)
```

```python
smiles = df['SMILES'].values
p_sat_kpa = 10**(-3) * df['pSat_Pa'].values 
log_p_sat_kpa = np.log10(p_sat_kpa)
with open('smiles.txt', 'w+') as file:
    for smi in smiles:
        file.writelines(smi + '\n')
np.savetxt('log_p_sat.txt', log_p_sat_kpa, fmt='%.15f')
```

### Summary of modifications and adaptations:
#### - Convert 'pSat_Pa' values to log10(kPa) units
#### - Save columns: 'SMILES', 'pSat_Pa' to files of different format 'smiles.txt', 'log_p_sat.txt' without headers
