```python
import pandas as pd 
import numpy as np
```

```python

data = pd.read_csv('VOC-Database.csv')
for x in data.columns:
    print(x)
data.head(5)
```

```python
smiles = data['SMILES'].values.astype(str)
dvap = data['dvap'].values.astype(str)
with open('smiles.txt', 'w+') as f:
    for smi in smiles:
        f.writelines(f'{smi}\n')
with open('dvap.txt', 'w+') as f:
    for el in dvap:
        f.writelines(f'{el}\n')
```

```python
train_data = data[data['VOC'] == 'NO']
test_data = data[data['VOC'] == 'YES']
print(len(train_data), len(test_data))
```

```python
smiles = train_data['SMILES'].values.astype(str)
dvap = train_data['dvap'].values.astype(str)
with open('smiles_custom_train.txt', 'w+') as f:
    for smi in smiles:
        f.writelines(f'{smi}\n')
with open('dvap_custom_train.txt', 'w+') as f:
    for el in dvap:
        f.writelines(f'{el}\n')
```

```python
smiles = test_data['SMILES'].values.astype(str)
dvap = test_data['dvap'].values.astype(str)
with open('smiles_custom_test.txt', 'w+') as f:
    for smi in smiles:
        f.writelines(f'{smi}\n')
with open('dvap_custom_test.txt', 'w+') as f:
    for el in dvap:
        f.writelines(f'{el}\n')
```

```python
data = pd.read_csv('VOC-Database.csv')
data = data.drop(columns=['CAS', 'dvap', 'SMILES', 'num', 'External', 'Family',
                   'Key', 'VOC'])
data.to_csv('EnsembleFC.txt', sep=' ', index=None, header=None)
```
