```python
import pandas as pd 
import numpy as np
```

```python
# Load the original data (Wang et al., 2017)
df = pd.read_csv('Supporting information CSV.csv', skiprows=1, sep=',')
df.head(3)
```

```python
# Convert P_sat to log10 units
df['COSMOtherm vapor pressure (kPa)'] = np.log10(df['COSMOtherm vapor pressure (kPa)'].values)

# Select the needed columns
cols = ['SMILES', 'COSMOtherm vapor pressure (kPa)', 'COSMOtherm_logKWIOM/G', 'COSMOtherm_logKW/G']
df = df[cols]
df.head(2)
```

```python
# Save the columns to separate data files
col_to_fname= {
    'SMILES': 'smiles.txt',
    'COSMOtherm vapor pressure (kPa)': 'log_p_sat.txt',
    'COSMOtherm_logKWIOM/G' : 'log_kwiomg.txt',
    'COSMOtherm_logKW/G' : 'log_kwg.txt'
}
for col, fname in col_to_fname.items():
    vals = df[col].values
    if col == 'SMILES':
        with open(fname, 'w+') as file:
            for val in vals:
                file.writelines(val + '\n')
    elif col == 'COSMOtherm vapor pressure (kPa)':
        np.savetxt(fname, vals, fmt='%.15f')
    else:
        np.savetxt(fname, vals, fmt='%.2f')
```

### Summary of modifications and adaptations:
#### - Convert COSMOtherm vapor pressure (kPa) values to log10(kPa) units
#### - Save columns: 'SMILES', 'COSMOtherm vapor pressure (kPa)', 'COSMOtherm_logKWIOM/G', 'COSMOtherm_logKW/G' to files of different format 'smiles.txt', 'log_p_sat.txt', 'log_kwiomg.txt', 'log_kwg.txt' without headers
