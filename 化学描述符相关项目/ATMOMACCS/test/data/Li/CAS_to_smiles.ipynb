```python
import pandas as pd
import cirpy as c
import numpy as np
```

```python
# Load first data file (CAS)
data = pd.read_csv('li_1.csv')
data['CAS'] = data['CAS'].apply(lambda x: x.lstrip('0')) # Remove leading zeros from CAS-numbers
data.head(5)
```

```python
# Convert CAS-numbers to SMILES strings using cirpy CAS resolver
cas_vals = data['CAS'].values
smiles = []
for cas in cas_vals:
    try:
        smiles.append(c.resolve(cas, 'smiles'))
    except:
        print('Error')

smiles = np.array(smiles)
data['SMILES'] = smiles
```

```python
data = data.drop(columns=['CAS']) # Remove unnecessary column
# Load second data file
data2 = pd.read_csv('li_2.csv')
```

```python
# Concatenate both data file contents to one dataframe
data_all = pd.concat([data, data2])
```

```python
data_all.values.shape
print(len(smiles[smiles == None]))
```

```python
data_all.to_csv('li_all.csv')
```

```python
# Prune unresolved data entries
data_no_nones = data[smiles != None]
tg = data_no_nones['Estimated Tg (K)'].values
data_no_nones = data_no_nones[tg != None]
```

```python
data_all_smiles = pd.concat([data_no_nones, data2])
data_all_smiles.to_csv('li_all_smiles.csv')
```

```python
# Save smiles strings
with open('smiles.txt', 'w+') as file:
    for smi in data_all_smiles['SMILES'].values:
        file.writelines(smi + '\n')
```

```python
data_smiles = pd.read_csv('li_all_smiles.csv')
data_smiles.head(10)
```

```python
data_smiles = data_smiles.dropna(subset=['Estimated Tg (K)'])
data_smiles.head(10)
```

```python
try:
    data_smiles = data_smiles.drop(columns=['Unnamed: 0', 'Unnamed: 0.1'])
    data_smiles.to_csv('li_all_smiles_cleaned.csv')
except:
    print('Something went wrong')
# Save smiles strings
with open('smiles.txt', 'w+') as file:
    for smi in data_smiles['SMILES'].values:
        file.writelines(smi + '\n')
# Save tg values
with open('tg.txt', 'w+') as file:
    for tg in data_smiles['Estimated Tg (K)'].values:
        file.writelines(
            tg.replace(',', '.') + '\n'
            )
```
